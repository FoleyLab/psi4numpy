## This tests the QED-RHF gradient components against Hilbert's and against numerical components


```python
from oo_cqed_rhf import CQEDRHFCalculator
import numpy as np
import psi4
psi4.core.be_quiet()


### Expected gradients for the backwards displaced O along z
### using our code for all terms in the gradient
_expected_grad_b_psi4_false = np.array(
[[-1.15058861e-16, -8.42176804e-18, -6.22261790e-02],
 [-1.38525048e-17, -4.37161985e-02,  3.11130895e-02],
 [ 1.28911365e-16,  4.37161985e-02,  3.11130895e-02]]
)

### using psi4 with density fitting for the canonical RHF gradient terms
_expected_grad_b_psi4_true = np.array(
[[ 2.25021410e-16,  3.71157596e-06, -6.22221403e-02],
 [-1.24759369e-16, -4.37167489e-02,  3.11127927e-02],
 [-1.00262040e-16, 4.37130373e-02,  3.11093476e-02]]
)

### Expected gradients for the forwards displaced O along z
### using our code for all terms in the gradient
_expected_grad_f_psi4_false = np.array(
[[ 1.78480435e-16, -1.21543163e-16, -6.14182262e-02],
 [-2.46917070e-16, -4.34548055e-02,  3.07091131e-02],
 [ 6.84366355e-17,  4.34548055e-02,  3.07091131e-02]]
)

### using psi4 with density fitting for the canonical RHF gradient terms
_expected_grad_f_psi4_true = np.array(
[[ 1.78480435e-16, -1.37205383e-16, -6.14182262e-02],
 [-2.46917070e-16, -4.34548055e-02,  3.07091131e-02],
 [ 6.84366355e-17,  4.34548055e-02,  3.07091131e-02]]
)
```

```python

## backwards displaced geometry string
h2o_string_b = """
         O            0.000000000000     0.000000000000    -0.069000000000
         H            0.000000000000    -0.790689573744     0.543701060715
         H            0.000000000000     0.790689573744     0.543701060715
no_reorient
nocom
symmetry c1
"""

## forward displaced geometry string
h2o_string_f = """
         O            0.000000000000     0.000000000000    -0.068000000000
         H            0.000000000000    -0.790689573744     0.543701060715
         H            0.000000000000     0.790689573744     0.543701060715
no_reorient
nocom
symmetry c1
"""


# lambda vector along z
lambda_vector = np.array([0, 0, 0.05])


# psi4 options
psi4_options = {
    "basis": "cc-pVDZ",
    "save_jk": True,
    "scf_type": "pk",
    "e_convergence": 1e-12,
    "d_convergence": 1e-12,
}



```

```python
# instantiate, using forward displaced geometry string... shouldn't matter because
# we will pass our desired geometry string when we want to compute the gradient
calc = CQEDRHFCalculator(lambda_vector, h2o_string_f, psi4_options)

# calculate the CQED-RHF energy and gradient at h2o_string_b, use our routines for all terms
qed_rhf_energy, qed_rhf_grad, g = calc.calc_force_and_energy(h2o_string_b, use_psi4_scf_grad=False)

# test to see if we are close
print(F"\n Pure QED-RHF gradient at backwards displaced geometry passes?\n {np.allclose(qed_rhf_grad, _expected_grad_b_psi4_false, 1e-6, 1e-6)}")

# calculate the CQED-RHF energy and gradient at h2o_string_b, use psi4 with density fitting for canonical terms
qed_rhf_energy, qed_rhf_grad, g = calc.calc_force_and_energy(h2o_string_b, use_psi4_scf_grad=True)

# test to see if we are close
print(F"\n Hybrid Pure-DF QED-RHF gradient at backwards displaced geometry passes?\n {np.allclose(qed_rhf_grad, _expected_grad_b_psi4_true, 1e-6, 1e-6)}")

# calculate the CQED-RHF energy and gradient at h2o_string_f, use our routines for all terms
qed_rhf_energy, qed_rhf_grad, g = calc.calc_force_and_energy(h2o_string_f, use_psi4_scf_grad=False)

# test to see if we are close
print(F"\n Pure  QED-RHF gradient at forward displaced geometry passes?\n {np.allclose(qed_rhf_grad, _expected_grad_f_psi4_false, 1e-6, 1e-6)}")

# calculate the CQED-RHF energy and gradient at h2o_string_b, use psi4 with density fitting for canonical terms
qed_rhf_energy, qed_rhf_grad, g = calc.calc_force_and_energy(h2o_string_b, use_psi4_scf_grad=True)

# test to see if we are close
print(F"\n Hybrid Pure-DF QED-RHF gradient at forward displaced geometry passes?\n {np.allclose(qed_rhf_grad, _expected_grad_f_psi4_true, 1e-6, 1e-6)}")

```
